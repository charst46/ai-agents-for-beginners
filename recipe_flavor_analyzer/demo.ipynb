# Recipe Flavor Analyzer - Demo

A rule-based flavor engine that analyzes recipes for balance across six dimensions:
**salt, fat, acid, sweet, umami, heat**

It detects imbalances and suggests fixes, with optional pantry matching and dietary constraints.

```python
import sys
sys.path.insert(0, '..')

from recipe_flavor_analyzer import analyze_recipe
import json

def show(result):
    print(json.dumps(result, indent=2))
```

## Example 1: Fat-heavy pasta (needs acid)

```python
recipe = """
- 3 tbsp butter
- 2 cups heavy cream
- 1 lb pasta
- 1 cup parmesan cheese
- salt
- black pepper
"""

result = analyze_recipe(recipe)
show(result)
```

## Example 2: With pantry items

```python
result = analyze_recipe(
    recipe,
    pantry_items=["lemon", "olive oil", "vinegar", "chili flakes"]
)
show(result)
```

## Example 3: With dietary constraints

```python
result = analyze_recipe(
    recipe,
    pantry_items=["lemon", "olive oil"],
    constraints=["no_dairy"]
)
show(result)
```

## Example 4: Well-balanced stir fry

```python
balanced_recipe = """
- chicken
- soy sauce
- lime juice
- sesame oil
- honey
- garlic
- ginger
- chili flakes
- salt
"""

result = analyze_recipe(balanced_recipe)
show(result)
```

## Example 5: Sweet dessert (needs acid)

```python
dessert = """
- 2 cups sugar
- 1 cup butter
- 3 cups flour
- 2 eggs
- 1 cup cream
- vanilla
- chocolate
"""

result = analyze_recipe(dessert)
show(result)
```
